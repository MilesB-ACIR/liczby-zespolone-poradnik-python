
# Liczby zespolone i ich reprezentacja graficzna — poradnik w Pythonie

Ten notatnik to krótki przewodnik po liczbach zespolonych: podstawy, operacje, forma trygonometryczna i geometryczna interpretacja (Argand), a także proste wizualizacje w Pythonie z użyciem `numpy` i `matplotlib`.


```

import numpy as np
import cmath
import math
import matplotlib.pyplot as plt

# Dla przejrzystości rysunków
plt.rcParams["figure.figsize"] = (6, 6)
plt.rcParams["figure.dpi"] = 120
```


## 1. Podstawy liczb zespolonych w Pythonie

- Jednostka urojona to `1j`.
- Liczba zespolona ma część rzeczywistą (`z.real`) i urojoną (`z.imag`).
- Moduł `abs(z)` zwraca |z|.
- Argument (fazę) można uzyskać przez `cmath.phase(z)`.


```

z = 3 + 4j
w = -1 + 2j

print("z =", z)
print("w =", w)
print("Re(z) =", z.real, " Im(z) =", z.imag)
print("|z| =", abs(z))
print("arg(z) =", cmath.phase(z))

# Operacje arytmetyczne
print("z + w =", z + w)
print("z * w =", z * w)
print("z / w =", z / w)

# Sprzężenie zespolone
print("sprzężenie z̄ =", z.conjugate())
```


## 2. Postać trygonometryczna i wzór Eulera

Każde `z ≠ 0` można zapisać jako:
\[ z = r(\cos\varphi + i\sin\varphi) = r\,e^{i\varphi} \]
gdzie \( r = |z| \) to moduł, a \( \varphi = \arg(z) \) to argument.


```

z = 1 - 1j
r = abs(z)
phi = cmath.phase(z)

print("z =", z)
print("r =", r)
print("phi (rad) =", phi)
print("Euler: r*e^{i phi} =", r * cmath.exp(1j * phi))
```


## 3. Rysowanie na płaszczyźnie zespolonej (Arganda)

Poniżej tworzymy pomocniczą funkcję do rysowania punktów/strzałek reprezentujących liczby zespolone.


```

def plot_complex(points, arrows=None, title="Płaszczyzna zespolona"):
    re_vals = [z.real for z in points]
    im_vals = [z.imag for z in points]

    plt.figure()
    plt.axhline(0, linewidth=1)
    plt.axvline(0, linewidth=1)
    plt.scatter(re_vals, im_vals, marker="o")
    if arrows:
        for (z_from, z_to) in arrows:
            plt.plot([z_from.real, z_to.real], [z_from.imag, z_to.imag], linewidth=1)
    plt.gca().set_aspect('equal', 'box')
    plt.grid(True, linestyle="--", linewidth=0.5)
    plt.xlabel("Re")
    plt.ylabel("Im")
    plt.title(title)
    plt.show()
```


### Przykład: losowe punkty zespolone


```

np.random.seed(0)
pts = (np.random.uniform(-2,2,10) + 1j*np.random.uniform(-2,2,10))
plot_complex(list(pts), title="Losowe punkty na płaszczyźnie zespolonej")
```


### Mnożenie jako rotacja i skalowanie

Mnożenie przez \( e^{i\theta} \) odpowiada **obrotowi** o kąt \( \theta \).  
Mnożenie przez liczbę rzeczywistą \( a>0 \) odpowiada **skalowaniu** długości wektora.


```

theta = math.pi/6
scale = 1.5
z = np.array([1+0.3j, -0.7+0.5j, 0.2-1.0j, -1.2-0.8j])
w1 = z * 1j
w2 = z * scale * np.exp(1j * theta)

plot_complex(list(z), arrows=list(zip(z, w1)), title="Rotacja o 90° (mnożenie przez i)")
plot_complex(list(z), arrows=list(zip(z, w2)), title="Skalowanie i rotacja (1.5·e^{iπ/6})")
```


### \(n\)-te pierwiastki z jedności

Punkty \( e^{2\pi i k/n} \) dla \( k=0,\dots,n-1 \) leżą na **okręgu jednostkowym** i są rozłożone równomiernie co \( 2\pi/n \).


```

n = 12
roots = np.exp(2j*np.pi*np.arange(n)/n)

plt.figure()
plt.axhline(0, linewidth=1); plt.axvline(0, linewidth=1)
plt.scatter(roots.real, roots.imag)
for r in roots:
    plt.plot([0, r.real], [0, r.imag], linewidth=0.8)
circle = np.exp(2j*np.pi*np.linspace(0,1,361))
plt.plot(circle.real, circle.imag, linewidth=0.8)
plt.gca().set_aspect('equal', 'box')
plt.xlabel("Re"); plt.ylabel("Im")
plt.title(f"{n}-te pierwiastki z jedności")
plt.grid(True, linestyle="--", linewidth=0.5)
plt.show()
```
